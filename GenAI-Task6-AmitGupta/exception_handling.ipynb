```python
#Task-1  Safe division utility
try:
    numerator = int(input("enter value for numerator"))
    denominator = int(input("enter value for denominator")) 
    result = numerator/denominator
except ValueError:
    print("input is not a number")
except ZeroDivisionError:
    print("denominator should not be zero")

# if try run successfully..
else:
    print(f"final result : {result:.2f}")

# always execute finally block
finally:
    print("Operation Complete.")
```

```python
# Task-2 Bill calculator with error handling.
prices = [120, 150, 350, 'abc', 500, -200, 800 ]
total=0
for price in prices:
    try:
        if not isinstance(price, (int,float)):
             raise TypeError("Value is not a number")
       
        if price < 0:
            raise ValueError("Negative price not allowed")
                
        total +=price

    except TypeError as e:
        print(f"TypeError: {e} ({price})")
        
    except ValueError as e:
        print(f"ValueError: {e} ({price})")

print(f"Total: {total}")
```

```python
# Task-3 custom exception for Age Validator
def check_age(age):
    if not age > 1 and age < 120:
        raise ValueError("Age must be between 1 and 120")
try:
    user_input = int(input("enter your age: "))
    check_age(user_input)
except ValueError as e:
    print(f"ValueError: {e}")
else:
    print(f"you are {user_input} years old!")   
```

```python
# Task-4 File render with exception handling
import os
try:
    file_path = "sensitive_data.txt"
    file = input("enter file name with extention")
   
    if file_path == file:
            raise PermissionError(f"Access Denied: You don't have permission to read this file")
    
    if not os.path.exists(file):
        raise FileNotFoundError(f"{file} file not found!")
    
except FileNotFoundError as e:
    print(f"FileNotFoundError: {e}")
except PermissionError as e:
    print(f"PermissionError: {e}")
else:
    with open(file, 'r') as f:
        print(f"file content: \n{f.read()}")
finally:
    print("File operation attempted")
```

```python
# Task-5 mini program: safe shoping cart..
cart = []

while True: 
    try:
        user_input = input("enter price: ")

        if user_input == 'q':
            break

        price = float(user_input)
       
        if price < 0 :
            raise ValueError("price should not be negative")
       
        if not isinstance(price, float):
            raise TypeError("invalid price entered")
        
        cart.append(price)
        print(price)
    
    except ValueError as e:
        print(f"ValueError: {e}")

    except TypeError as e:
            print(f"TypeError: {e}")

print(f"Total bill: {sum(cart)}")
print(f"Total items: {len(cart)}")
```
