```python
class Product:
    def __init__(self, name, price, category):
        self.name = name
        self.category = category
        self.price = price

    def get_info(self):
        print(
            f"product name: {self.name} \ncategory: {self.category} \nprice: {self.price}"
        )
        print("---" * 10)

    def apply_discount(self, percent):
        discounted_price = self.price - (self.price * percent)
        print(f"Discounted price: {discounted_price:.2f}")


product_1 = Product("car", 1000, "four-whiler")
product_2 = Product("bike", 500, "two-whiler")

product_1.get_info()
product_2.get_info()

product_1.apply_discount(0.10)
```

```python
# Task-2 Constructor and Encapsulation
class Product:
    def __init__(self, name, price, category):
        self.name = name
        self.category = category
        self.__price = price

    def get_price(self):
        return self.__price

    def set_price(self, new_price):
        if new_price > 0 :
            self.__price = new_price
        else:
            print("price should be greater than zero")
product = Product("shirt", 1000, "clothes")

print(f"before update: {product.get_price()}")
product.set_price(3500)
print(f"after update: {product.get_price()}")
```

```python
# Task-3 Inheritance (Single Level)
class Product:
    def __init__(self, name, price, category):
        self.name = name
        self.category = category
        self.price = price

    def get_info(self):
        print(
            f"product name: {self.name} \ncategory: {self.category} \nprice: {self.price}"
        )
        print("---" * 10)

class ElectronicProduct(Product):
    def __init__(self, name, price, category, warranty_year):
        super().__init__(name, price, category)
        self.warranty_year = warranty_year

    def get_info(self):
        super().get_info()
        print(f"warranty year: {self.warranty_year}")
        print("---" * 10)

product = Product('mobile', 5000, 'electronics')
electronic_product = ElectronicProduct("bottle", 2026, 'utility', 2026)

product.get_info()
electronic_product.get_info()
```

```python
# Task-4  Polymorphism
class Laptop(Product):

    def get_info(self):
        print(f"Laptop Model: {self.name}, \nPrice: {self.price} and \ncategory of laptop: {self.category}")

class Mobile(Product):

    def get_info(self):
        print(f"Mobile model: {self.name}, \nprice of mobile:{self.price} and \ncategory of mobile: {self.category}")

laptop = Laptop("Lenovo Ideapad-3", 45000, 'lenovo')
mobile = Mobile("Samsun Galaxy S5", 85000, 'samsung')

for obj in [laptop, mobile]:
    obj.get_info()
    print("="*40)
```

```python
# Task-6  Magic Methods and Operator Overloading..
class Product:
    def __init__(self, name, price, category):
        self.name = name
        self.category = category
        self.price = price

    def __str__(self):
        return f"product name: {self.name}, price: {self.price}, category: {self.category}"

    def __add__(self, other):

        if isinstance(other, Product):
            return self.price + other.price

        if isinstance(other  (int, float)):
            return self.price + other
    
        return NotImplemented

product1 = Product("BMW- S454", 35000, "BMW")
product2 = Product("Audi-Q7", 50000, "Audi")

total_price = product1 + product2
print(f"Total Price: {total_price}")
print(product1)
print(product2)
```

```python
# Task-7  Simple Inventory System (OOP Only)
class Inventory:
    
    def __init__(self):
       self.products =[]

    def add_product(self, product):
        self.products.append(product)
        print(self.products)

    def is_empty(self):
        return len(self.products) == 0
    
    def remove_product(self, name):

        for product in self.products:
            if product["name"] == name:
                self.products.remove(product)
                print(f" {name} removed successfully")
                return
        print(f"{name} not found")
    
    def get_total_value(self):
        total = 0
        if self.is_empty():
            raise ValueError("Product list is empty")
        for product in self.products:
            total += product["price"]
        return total

    def show_all_products(self):
        for product in self.products:
            print(f"Product: {product['name']}, Price: {product['price']}")

class Store:
    def __init__(self, store_name):
        self.store_name = store_name
        self.inventory = Inventory()

    def add_new_product(self):
        product_name =  input("enter product name")
        product_price = int(input("enter product price"))

        if product_price <= 0: 
            raise ValueError("product price should be > 0")
        new_product_details = {'name':product_name.strip(), 'price':product_price}
        self.inventory.add_product(new_product_details)

    def show_summary(self):
        print(f"\nTotal items: {len(self.inventory.products)}")
        print(f"\nTotal values: {self.inventory.get_total_value()}")
        print("All Products: \n")
        self.inventory.show_all_products()

try:
    store= Store('FlipKart')

    store.add_new_product()
    store.add_new_product()
    store.add_new_product()

    store.show_summary()

    if len(store.inventory.products) >= 2:
        p1 = store.inventory.products[0]
        p2 = store.inventory.products[1]
        print(f"combined price: {p1['price']+p2['price']}")
    
except ValueError as e:
    print(f"ValueError: {e}")
except TypeError as e:
    print(f"TypeError: {e}")
```
