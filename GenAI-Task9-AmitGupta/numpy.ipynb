```python
# Task-1 Creating NumPy Arrays.
import numpy as np

# 1D Array
oneDArray = np.arange(1,10, dtype=np.int8)
print(oneDArray)
print(f"Shape of oneDArray: {oneDArray.shape}")
print(f"Data Type of oneDArray: {oneDArray.dtype}")
print("="*40)

# 2D Array
twoDArray = np.arange(1,10, dtype=np.int8).reshape(3,3)
print(twoDArray)
print(f"Shape of twoDArray: {twoDArray.shape}")
print(f"Data Type of twoDArray: {twoDArray.dtype}")
print("="*40)

my_list = [10,20,30,40,50]
numpy_array = np.array(my_list, dtype=np.int8)   ### range of values we can store with in int8 = -128 to 127
print(numpy_array)
print(f"Shape of numpy_array: {numpy_array.shape}")
print(f"Data Type of numpy_array: {numpy_array.dtype}")
print("="*40)
```

```python
# Task-2  Mathematical Operation..

A = np.array([10,20,30,40])
B = np.array([1,2,3,4])

print(f"Addition: {A+B}")
print(f"Subtraction: {A-B}")
print(f"Multiplication: {A*B}")
print(f"Division: {A/B}")
print(f"Power: {A**2}")

print("==== using numpy functions ====")
print(f"Addition: {np.add(A,B)}")
print(f"Subtraction: {np.subtract(A, B)}")
print(f"Multiplication: {np.multiply(A, B)}")
print(f"Division: {np.divide(A, B)}")
print(f"Power: {np.power(A,2)}")
```

```python
# Task-3  NumPy Mathematical Formulas

values = np.array([2,4,6,8,10])

print(f"Square root of each value: {np.sqrt(values)}")

# Exponentional of each value
print(f"Exponentional of each value: {np.round(np.exp(values),4)}")

# natural logarithm..
print(f"Natural logarithm of each element: {np.round(np.log(values),4)}")

## sum of all values
print(f"Sum of all the values: {np.sum(values)}")

## Cumulative sum of values
print(f"Cumulative sum of values: {np.cumsum(values)}")
```

```python
# Task-4  Aggregation Functions...
data = np.arange(10,100, 10, dtype=np.int16).reshape(3,3)

print(data)

# row-wise  and column-wise sum
print(f"Row-wise sum: {np.sum(data, axis=1)}")
print(f"Column-wise sum: {np.sum(data, axis=0)}")

# min and max value using np.min(), np.max()
print(f"Min value: {np.min(data)}")
print(f"Max value: {np.max(data)}")

# Overall mean..
print(f"Mean : {np.mean(data)}")  ## avg of all the values of array..
```

```python
## Task-5  Statistical Operations..

marks = [75, 68, 61, 99, 56, 65, 84, 74, 63, 72, 91] 
# np.random.randint(50,101, size=11)

print(f"Marks = {marks}")
# Mean
print(f"Mean: {np.mean(marks)}")
# Median
print(f"Median= {np.median(marks)}")

# standard deviation
print(f"Variance: {np.var(marks):.2f}")

#Varance
print(f"Standard Deviation: {np.std(marks):.2f}")

# Min and Max
print(f"Min = {np.min(marks)}")
print(f"Max = {np.max(marks)}")

# Range
print(f"Range = {np.max(marks) - np.min(marks)}")
```

```python
# Task-6  Percentiles and Sorting..

print(f"Marks array: {marks}")

print(f"Mean: {np.round(np.mean(marks))}")
print(f"Sorted array: {np.sort(marks)}")
print(f"Find 25th percentile: :{np.percentile(marks, 25)}")
print(f"Find 50th percentile: :{np.percentile(marks, 50)}")
print(f"Find 75th percentile: :{np.percentile(marks, 75)}")

## marks above avg marks
print(f"Marks above avg marks: {np.sum(marks > np.mean(marks))}")  ## > symbol will return True and False and applied sum add them..
```

```python
# Task-7  Mini Use Cases  Sales Analysis..

sales = np.array( [1200, 1500, 900, 2000, 1800, 1700, 1600])  ## here 0 index is 1st day...

avg_sales = np.mean(sales)

print(f"daily sales: {sales}")
print(f"Total weekly sales: {np.sum(sales)}")

print(f"Average daily sales: {np.round(avg_sales)}")

print(f"Highest sales Day: {np.argmax(sales)+1} and max sale is: {np.max(sales)}")
print(f"Lowest sales Day: {np.argmin(sales)+1} and min sale is: {np.min(sales)}")

print(f"Standard Deviantion: {np.round(np.std(sales), 4)}")

print(f"days where sales were above average: {np.where(sales > avg_sales)[0]+1 }")
# keeping only True elements from sales..
print(f"Sales above average : {sales[sales > avg_sales]}")
```
