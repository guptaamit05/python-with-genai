```python
### sum()

ar = np.round(np.random.random((4,4))*10).astype(np.int32)
ar
```

```python
sum_col_each = np.sum(ar, axis=0)  ## axis=0 sum each column of matrix
print(sum_col_each)

sum_row_each = np.sum(ar, axis = 1)    ## axis=1 means sum each row of matrix
print(sum_row_each)
```

```python
prod_ar = np.prod(ar)
print(prod_ar)

row_prod = np.prod(ar, axis=1)  ## row wise
print(row_prod)

col_prod = np.prod(ar, axis=0)  # column wise
print(col_prod)
```

```python
# Statistical Function in NumPy
mean_row = np.mean(ar, axis=1)
mean_col = np.mean(ar, axis=0)

print(f"row mean: {mean_row}")
print(f"column mean: {mean_col}")
```

```python
arr_mean = ar.mean()
print(f"array mean: {arr_mean}")
```

```python
### nditer()  function 

arr_tt = np.arange(1,28).reshape(3,3,3)
for i in np.nditer(arr_tt):
    print(i)
```

```python
### Stacking and Splitting...

# Stacking:
ar1= np.arange(1,13).reshape(3,4)
ar2 = np.arange(13,25).reshape(3,4)
print(ar1)
print(ar2)

h_stacking = np.hstack((ar1,ar2))
print(h_stacking)
```

```python
ar1= np.arange(1,13).reshape(3,4)
ar2 = np.arange(13,25).reshape(3,4)
print(ar1)
print(ar2)
print("=== result=======")

v_stacking = np.vstack((ar1,ar2))
print(v_stacking)
```

```python

np.hsplit(v_stacking, 2)
```

```python
np.vsplit(v_stacking, 3)
```
