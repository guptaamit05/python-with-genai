```python
# Task-1  (Write sales order in file)
sales = [1200, 450, 980, 1500, 3000]

try:
    with open("sales_data.txt", 'w') as f:
        #f.write(",".join(str(sale) for sale in sales))
        f.write("\n".join(str(sale) for sale in sales))
    with open("sales_data.txt", 'r') as f:
        print(f.read())

except Exception as e:
    print(f"Error: {e}")
```

```python
# Task-2 Read the same file with read(), readline() and readlines() function.
try:
    with open('sales_data.txt', 'r') as f:
        # with read()
        print(f"with read() function: {f.read()}")
        print("--"*20)
        f.seek(0)
        # with readline()
        while True:
            line = f.readline()
            if not line:
                break
            print(f"reading file using readline() : {line}")

        # with readlines()
        f.seek(0)
        print("--"*20)
        all_lines= f.readlines()
        print(f"using readlines() function: {list(int(line.strip()) for line in all_lines)}")
        print("--"*20)

        # with readline()
except Exception as e:
    print(f"Error: {e}")
```

```python
# Task-3 Append New sales
new_sales = [5000, 2500, 1700]
try:
    with open('sales_data.txt', 'a+') as f:

        if(f.tell()>0):
            f.seek(f.tell()-1)
            if f.read(1) != "\n":
                f.write("\n")
        f.write("\n".join(str(sale) for sale in new_sales))

    with open('sales_data.txt', 'r') as f:
        print(f"udpated file content using append(): {f.read()}")
        # reset the cursor the the start of file
        f.seek(0)
        print(f"Total numbers of lines: {len(f.readlines())}")

except Exception as e:
    print(f"Error: {e}")
```

```python
# Task-4 generate summary report from file...
all_sales=[]
try:
    with open('sales_data.txt', 'r') as f:
        while True:
            line = f.readline()
            if not line:
                break
            remove_slash_n = line.replace("\n", "")
            all_sales.append(remove_slash_n)
except Exception as e:
    print(f"Error:{e}")

# print all sales value
print(all_sales)
try:
    int_sales = [int(sale) for sale in all_sales]
    # integer sales
    print(f"sales with integer values: {int_sales}")
    # calculate and print:
    print(f"Total Sales: {len(int_sales)}")
    print(f"Highest Sale: {max(int_sales)}")
    print(f"Lowest Sale: {min(int_sales)}")

    print(f"Average Sale: {sum(int_sales)/len(int_sales)}")
except ZeroDivisionError as e:
    print(f"ZeroDivisionError: {e}")
except Exception as e:
    print(f"Error: {e}")
```

```python
# Task-5  create product info file  (user input)

def write_to_file(filename, content):
    try:
        with open(filename, 'a+') as f:
            if (f.tell()>0):
                f.seek(f.tell()-1)  # setting cursor to the second last position and check if it is == \n
                if (f.read(1) != '\n'):   # read single character and comparing it is \n or not
                    f.write("\n")
            f.write(content)
    except Exception as e:
        print(f"something went wrong: {e}")

try:
    for x in range(1,4):
        product_name = input(f"enter {x} product name")
        product_price = int(input(f"enter {x} product price"))
        write_to_file("product.txt", f"{product_name} | {product_price}")

    with open('product.txt', 'r') as f:
        while True:
            line = f.readline()
            if not line:
                break
            print(line)
except Exception as e:
    print("Error: {e}")
```

```python
# Task-7 Mini Project - Export Discounted Prices
prices = {
    "Mouse": 500,
    "Keyboard":800,
    "Monitor":7000,
    "Camera":5000,
    "Pendrive":400
}

filename = "discount_report.txt"
try:
    user_input = int(input('enter discount % (ex: 5)'))
    discount = round(user_input/100,2)
    total_item = len(prices)
    sum_discounted_price =0
    for item,price in prices.items():
        discounted_price = price + (price*discount)
        write_to_file(filename, f"{item}|{price}|{discounted_price}")
        sum_discounted_price +=discounted_price

    print(f"Average Discounted Price: {sum_discounted_price/total_item}")
    with open(filename, 'r') as f:
        print(f"file content: {f.read()}")
        print("-"*20)
except Exception as e:
    print(f"Error: {e}")
```
